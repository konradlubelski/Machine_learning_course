```python
from ml_finance.data.generate import generate_normal_sample
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
sample = generate_normal_sample(
    seed=42
    ,sample_size=1000
    ,mean=0.0
    ,standard_deviation=1.0)

sample[:10]
```

```python
summary = pd.Series(sample).describe()
```

```python
summary
```

```python
figure_path = "../reports/figures/normal_distribution_histogram.png"
plt.figure(figsize=(10,6))

plt.hist(
    sample
    ,bins=30
    ,edgecolor="black")

plt.title("A sample of 1000 observations from the N(0, 1) distribution")
plt.xlabel("Value")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig(
    figure_path,
    dpi=150,
)

plt.show()
```

```python
sample_100 = generate_normal_sample(
    seed=42,
    sample_size=100,
)

sample_1000 = generate_normal_sample(
    seed=42,
    sample_size=1000,
)

sample_10000 = generate_normal_sample(
    seed=42,
    sample_size=10000,
)
```

```python
comparison = pd.DataFrame({
    "sample_size": [
        100
        ,1000
        ,10000
        ]
    ,"mean":[
        sample_100.mean()
        ,sample_1000.mean()
        ,sample_10000.mean()]
    ,"standard_deviation":[
        sample_100.std()
        ,sample_1000.std()
        ,sample_10000.std()]
})

comparison
```
